```python
##exCEPTION hANDLING##

##try,except,else,finally
try:
    x = int(input("Enter x : "))
    ans = 10/x

except ZeroDivisionError:
    print(f"Dvide by 0 is not allowed!")

except ValueError:
    print(f"Please enter a valid integer!")

else:
    print(f"ans = {ans }")
```
