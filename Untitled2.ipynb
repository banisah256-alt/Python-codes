```python
info = [
    ("Alice", "Math"),
    ("Bob", "Science"),
    ("Alice", "Science"),
    ("Charlie", "Math"),
    ("Bob", "Math"),
    ("Alice", "English"),
    ("Charlie", "English")
]

student_courses = {}

for name, course in info:
    if student_courses.get(name) == None:
        student_courses[name] = [course]
    else:
        student_courses[name].append(course)

print(student_courses)
```
